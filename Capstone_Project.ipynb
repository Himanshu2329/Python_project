# TASK 1: CREATE DATAFRAMES AND SAVE AS CSV

```python
# Importing basic libraries for calculation and drawing charts
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
# import matplotlib.pyplot as plt
# import seaborn as sns

# Making a dictionary for project data with missing values (np.nan)
project_dictionary = {
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005', 'A002', 'A005', 'A003', 'A001', 'A003', 'A001', 'A004', 'A004', 'A005'],
    'Project': ['Project 1', 'Project 2', 'Project 3', 'Project 4', 'Project 5', 'Project 6', 'Project 7', 'Project 8', 'Project 9', 'Project 10', 'Project 11', 'Project 12', 'Project 13', 'Project 14'],
    'Cost': [1002000, 2000000, 4500000, 5500000, np.nan, 680000, 400000, 350000, np.nan, 300000, 2000000, 1000000, 3000000, 200000],
    'Status': ['Finished', 'Ongoing', 'Finished', 'Ongoing', 'Finished', 'Failed', 'Finished', 'Failed', 'Ongoing', 'Finished', 'Failed', 'Ongoing', 'Finished', 'Finished']
}
# Converting to pandas dataframe and saving as csv
project_master = pd.DataFrame(project_dictionary)
project_master.to_csv('project.csv', index=False)

# Making a dictionary for employee info
employee_dictionary = {
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005'],
    'Name': ['John Alter', 'Alice Luxumberg', 'Tom Sabestine', 'Nina Adgra', 'Amy Johny'],
    'Gender': ['M', 'F', 'M', 'F', 'F'],
    'City': ['Paris', 'London', 'Berlin', 'Newyork', 'Madrid'],
    'Age': [25, 27, 29, 31, 30]
}
# Converting to dataframe and saving as csv
employee_master = pd.DataFrame(employee_dictionary)
employee_master.to_csv('employee.csv', index=False)

# Making a dictionary for seniority levels
seniority_dictionary = {
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005'],
    'Designation Level': [2, 2, 3, 2, 3]
}
# Converting to dataframe and saving as csv
seniority_master = pd.DataFrame(seniority_dictionary)
seniority_master.to_csv('seniority.csv', index=False)

print("Task 1 Done. All CSV files saved successfully.")
```

# Task 2 Compute missing costs via running average for loop

```python
# Reading the saved project csv file
my_project_df = pd.read_csv('project.csv')

# Using a for loop to check each row one by one
for index_num in range(len(my_project_df)):
    # Checking if the cost value is blank or NaN
    if pd.isna(my_project_df.loc[index_num, 'Cost']):
        # Calculating the running average up to this current row
        current_running_mean = my_project_df['Cost'].expanding().mean().iloc[index_num]
        # Replacing the missing value with the running average
        my_project_df.loc[index_num, 'Cost'] = current_running_mean

print("Task 2 Done. Missing costs are filled.")
```

# Task 3 Splitting Name

```python
# Reading the saved employee csv file
my_employee_df = pd.read_csv('employee.csv')

# Splitting the full name into two parts using space
split_data = my_employee_df['Name'].str.split(' ', n=1, expand=True)
my_employee_df['First Name'] = split_data[0]
my_employee_df['LastName'] = split_data[1]

# Removing the old 'Name' column from dataframe
my_employee_df.drop(columns=['Name'], inplace=True)

print("Task 3 Done. Name split into First Name and LastName.")
```

# Task 4: Joining All DataFrames into 'Final'

```python
# Reading the seniority csv file
my_seniority_df = pd.read_csv('seniority.csv')

# Joining employee and seniority table using ID column
first_merge = pd.merge(my_employee_df, my_seniority_df, on='ID')
# Joining the result with project table using ID column
total_merge_df = pd.merge(first_merge, my_project_df, on='ID')

# Naming the final combined dataframe as 'Final'
Final = total_merge_df.copy()

print("Task 4 Done. All dataframes joined into 'Final'.")
```

# Task 5: Adding 5% Bonus for Finished Projects

```python
# Creating a new column named Bonus with default value 0.0
Final['Bonus'] = 0.0

# If the status is 'Finished', calculate 5% bonus from the project cost
Final.loc[Final['Status'] == 'Finished', 'Bonus'] = Final['Cost'] * 0.05

print("Task 5 Done. 5% bonus added for finished projects.")
```

# Task 6: Demoting Failed Projects & Deleting Level > 4

```python
# If project status is Failed, increase designation level number by 1 (Demotion)
Final.loc[Final['Status'] == 'Failed', 'Designation Level'] += 1

# Removing records where designation level goes above 4
Final = Final[Final['Designation Level'] <= 4]

print("Task 6 Done. Failed projects demoted and filtered.")
```

# Task 7: Adding “Mr.” and “Mrs.” to the first name column and drop the gender column. 

```python
# Setting up simple rules: M means Mr. and F means Mrs.
prefix_rules = {'M': 'Mr. ', 'F': 'Mrs. '}

# Checking if Gender column exists before performing the operation
if 'Gender' in Final.columns:
    # Adding the prefix before the First Name column
    Final['First Name'] = Final['Gender'].map(prefix_rules) + Final['First Name']
    # Deleting the old gender column completely
    Final.drop(columns=['Gender'], inplace=True)
    print("Cell 8: Task 7 Done. Prefixes added and gender column dropped.")
else:
    # If already deleted, it will print this and skip without error
    print("Task 7 Already Processed. Gender column was already dropped.")
```

# Task 8: Promoting over 29 Age via IF Condition

```python
# Looping through each row of the Final dataframe
for row_id, current_row in Final.iterrows():
    # IF condition to check if the employee's age is more than 29
    if current_row['Age'] > 29:
        # Promoting means reducing the level number (e.g. from 3 to 2)
        Final.loc[row_id, 'Designation Level'] -= 1

print("Task 8 Done. Age-based promotions applied.")
```

# Task 9 : Total Project Cost

```python
# Grouping data by ID and First Name to add up all project costs
TotalProjCost = Final.groupby(['ID', 'First Name'])['Cost'].sum().reset_index()

# Renaming the columns as requested in task 9
TotalProjCost.columns = ['ID', 'First Name', 'Total cost']

print("-OUTPUT-")
print(TotalProjCost)
```
